```python
# patch:refvos\eval_ytvos_jf.py
"""
J&F baseline evaluation of the RefVOS pipeline on Refer-YouTube-VOS *valid*.

Refer-YouTube-VOS valid ships DENSE per-object GT in a nested layout:
    <valid_dir>/meta_expressions.json   {videos:{vid:{objects:{oid:{expressions,frames}}}}}
    <valid_dir>/JPEGImages/<vid>/<frame>.jpg
    <valid_dir>/Annotations/<vid>/<oid>/<frame>.png   (BINARY mask of object oid)

For each (video, object, expression):
  Module 1 MAttNet grounds the expression per frame -> boxes
  Module 2 temporal consistency picks one box / frame
  Module 3 segmenter -> binary mask
scored against the object's binary GT mask. Reports J (region IoU), F (DAVIS boundary),
J&F, averaged per expression then overall. Single-object task (no multi-object union),
so this is a fairer setting for MAttNet than MeViS.

Run from refvos/:
  python eval_ytvos_jf.py --valid_dir ../data/youtube-vos-2019/valid \
      --dataset_splitBy ytvos_train --max_videos 5 --frame_stride 2
"""
from __future__ import annotations
import os, os.path as osp, sys, json, time, argparse, warnings
import numpy as np
warnings.filterwarnings("ignore")

NB_DIR = os.getcwd()
if NB_DIR not in sys.path:
    sys.path.insert(0, NB_DIR)
import cv2
from PIL import Image
from refvos.io_data import load_rgb
from refvos.mattnet_grounder import build_module1, ground_frame
from refvos.temporal_consistency import Proposal, TemporalConsistency
from refvos.optical_flow import flow_magnitude_channel
from eval_mevis_jf import db_iou, db_boundary_f      # reuse the J&F metrics


def gt_mask(valid_dir, vid, oid, frame, layout):
    """Read the GT mask of object `oid` at `frame`.
    layout='perobj'  : Refer-YouTube-VOS valid -> Annotations/<vid>/<oid>/<frame>.png (binary)
    layout='palette' : Ref-DAVIS / ytvos train -> Annotations/<vid>/<frame>.png  (value==oid)
    """
    if layout == "palette":
        p = osp.join(valid_dir, "Annotations", vid, frame + ".png")
        if not osp.isfile(p):
            return None
        return (np.array(Image.open(p)) == int(oid)).astype(np.uint8)
    p = osp.join(valid_dir, "Annotations", vid, str(oid), frame + ".png")
    if not osp.isfile(p):
        return None
    return (np.array(Image.open(p)) > 0).astype(np.uint8)


def main(args):
    MATTNET_ROOT = osp.abspath("..")
    import torch
    dev = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    meta = json.load(open(osp.join(args.valid_dir, "meta_expressions.json")))["videos"]
    jpg_root = osp.join(args.valid_dir, "JPEGImages")

    tsf, mattnet, opt, w2i, mrcn, dev = build_module1(
        MATTNET_ROOT, model_id=args.model_id,
        dataset_splitBy=args.dataset_splitBy, device=dev)

    segmenter = None
    if not args.no_seg:
        from refvos.segmentation import build_model, BoxRefineSegmenter
        if args.refinement_checkpoint:
            model = build_model(pretrained=True, keep_pretrained_head=False)
            ckpt = torch.load(args.refinement_checkpoint, map_location="cpu")
            state = ckpt["model"] if isinstance(ckpt, dict) and "model" in ckpt else ckpt
            model.load_state_dict(state)
            print("loaded 5-channel refinement:", args.refinement_checkpoint)
        else:
            print("warning: no --refinement_checkpoint; using VOC demo head")
            model = build_model(pretrained=True, keep_pretrained_head=True)
        segmenter = BoxRefineSegmenter(model=model, device=dev)

    vids = list(meta)[:args.max_videos]
    all_J, all_F = [], []
    t0 = time.time()
    for vid in vids:
        objects = meta[vid]["objects"]
        for oid, obj in objects.items():
            frames = obj["frames"]
            sel = frames[::args.frame_stride]
            for ei, expr in enumerate(obj["expressions"][:args.max_exprs]):
                # ---- Module 1: ground per frame ----
                props, bgr_by_f = [], {}
                for fr in sel:
                    ip = osp.join(jpg_root, vid, fr + ".jpg")
                    if not osp.isfile(ip):
                        continue
                    bgr_by_f[fr] = cv2.imread(ip)
                    dets, per_q = ground_frame(tsf, mattnet, opt, w2i, mrcn, ip, [expr], dev)
                    fi = frames.index(fr)
                    for k, d in enumerate(dets):
                        props.append(Proposal(box=d["box"], score=float(per_q[0][k]),
                                              objectness=float(d["score"]), frame_idx=fi, det_id=k))
                # ---- Module 2: temporal rerank ----
                if props and args.temporal:
                    s_hat, _ = TemporalConsistency().rerank(props)
                else:
                    s_hat = np.array([p.score for p in props])
                ranked = {}
                for p, sc in zip(props, s_hat):
                    ranked.setdefault(p.frame_idx, []).append((sc, p))
                for fi in ranked:
                    ranked[fi].sort(key=lambda t: -t[0])

                # ---- Module 3 + J&F per frame ----
                Js, Fs = [], []
                for fr in sel:
                    gt = gt_mask(args.valid_dir, vid, oid, fr, args.gt_layout)
                    if gt is None:
                        continue
                    fi = frames.index(fr)
                    pred = np.zeros(gt.shape, np.uint8)
                    rk = ranked.get(fi, [])
                    if rk and segmenter is not None and fr in bgr_by_f:
                        rgb = load_rgb(osp.join(jpg_root, vid, fr + ".jpg"))
                        flow = None
                        if args.flow:
                            si = sel.index(fr)
                            prev = bgr_by_f.get(sel[si-1]) if si > 0 else None
                            nxt = bgr_by_f.get(sel[si+1]) if si+1 < len(sel) else None
                            flow = flow_magnitude_channel(prev, bgr_by_f[fr], nxt)
                        for _, p in rk[:args.topk]:
                            # A trained fg/bg refinement head has learned the box
                            # channel directly; the handcrafted demo prior is only
                            # for the untrained/VOC fallback.
                            prior = 0.0 if args.refinement_checkpoint else 4.0
                            m = segmenter.predict_mask(
                                rgb, list(p.box), flow,
                                box_prior_strength=prior,
                            )
                            if m.shape != gt.shape:
                                # Some Refer-YTVOS JPEGs carry EXIF orientation
                                # and/or use a different resolution from the
                                # released GT masks. Compare in GT coordinates.
                                m = cv2.resize(
                                    m.astype(np.uint8),
                                    (gt.shape[1], gt.shape[0]),
                                    interpolation=cv2.INTER_NEAREST,
                                )
                            pred = np.logical_or(pred, m).astype(np.uint8)
                    Js.append(db_iou(pred, gt)); Fs.append(db_boundary_f(pred, gt))

                if not Js:
                    continue
                J, F = float(np.mean(Js)), float(np.mean(Fs))
                all_J.append(J); all_F.append(F)
                print("  [%s o%s] %-42s J=%.3f F=%.3f J&F=%.3f"
                      % (vid[:8], oid, (expr[:40] + "..") if len(expr) > 42 else expr, J, F, (J + F) / 2))

    mJ = float(np.mean(all_J)) if all_J else 0.0
    mF = float(np.mean(all_F)) if all_F else 0.0
    print("\n=============== Refer-YouTube-VOS valid  J&F baseline ===============")
    print("checkpoint=%s | videos=%d (object,expr)=%d | stride=%d topk=%d temporal=%s flow=%s"
          % (args.dataset_splitBy, len(vids), len(all_J), args.frame_stride, args.topk, args.temporal, args.flow))
    print("J = %.1f%%   F = %.1f%%   J&F = %.1f%%   (%.0fs)"
          % (100*mJ, 100*mF, 100*(mJ+mF)/2, time.time()-t0))


if __name__ == "__main__":
    ap = argparse.ArgumentParser()
    ap.add_argument("--valid_dir", default="../data/youtube-vos-2019/valid")
    ap.add_argument("--dataset_splitBy", default="ytvos_train",
                    help="MAttNet checkpoint to ground with (refcoco_unc / ytvos_train / ...)")
    ap.add_argument("--model_id", default="mrcn_cmr_with_st",
                    help="checkpoint basename inside output/<dataset_splitBy>")
    ap.add_argument("--max_videos", type=int, default=5)
    ap.add_argument("--max_exprs", type=int, default=9999, help="expressions per object")
    ap.add_argument("--frame_stride", type=int, default=2)
    ap.add_argument("--topk", type=int, default=1)
    ap.add_argument("--gt_layout", choices=["perobj", "palette"], default="perobj",
                    help="perobj = Refer-YTVOS valid (nested binary); palette = Ref-DAVIS")
    ap.add_argument("--temporal", type=int, default=1)
    ap.add_argument("--flow", type=int, default=1)
    ap.add_argument("--no_seg", action="store_true")
    ap.add_argument("--refinement_checkpoint", default=None,
                    help="trained 5-channel fg/bg checkpoint; omit for VOC demo head")
    args = ap.parse_args()
    args.temporal = bool(args.temporal); args.flow = bool(args.flow)
    main(args)

```

```python
# refvos\refvos\io_data.py
"""
I/O utilities: dataset loading and the Module-1 -> Module-2 bridge.

Supports the Refer-YouTube-VOS / Refer-DAVIS directory layout (which is what
the provided sample.zip uses):

    <root>/
        meta_expressions.json
        JPEGImages/<video>/<frame>.jpg
        Annotations/<video>/<frame>.png      # palette PNG, value k = object id k

Two ways to obtain per-frame proposals for Module 2:

  (A) MAttNetJsonProposals  — the REAL path. Point it at the JSON your Module 1
      (test_single_frame.py, extended to dump per-frame detections) writes. No
      MAttNet/MaskRCNN code needs to run inside this package.

  (B) GTBoxProposals        — the DEMO path. Derives boxes from the ground-truth
      masks (+ synthetic jitter and a couple of distractor boxes) so the whole
      Module 2/3 pipeline can be exercised end-to-end without the heavy Module 1
      stack. This SIMULATES Module 1; it is not a substitute for it.
"""

from __future__ import annotations

import json
import os
import os.path as osp
from dataclasses import dataclass
from typing import Dict, List, Optional, Sequence, Tuple

import numpy as np
from PIL import Image, ImageOps

from .temporal_consistency import Proposal


# --------------------------------------------------------------------------- #
# Dataset
# --------------------------------------------------------------------------- #
@dataclass
class VideoData:
    video_id: str
    frame_names: List[str]                 # e.g. ["00000", "00005", ...]
    objects: Dict[str, dict]               # obj_id -> {"expressions": [...], ...}
    jpeg_dir: str
    ann_dir: Optional[str]                 # None if no GT available (test split)

    def frame_path(self, name: str) -> str:
        return osp.join(self.jpeg_dir, name + ".jpg")

    def ann_path(self, name: str) -> Optional[str]:
        if self.ann_dir is None:
            return None
        return osp.join(self.ann_dir, name + ".png")


def load_dataset(root: str) -> Dict[str, VideoData]:
    """Read meta_expressions.json and assemble VideoData for each video."""
    meta = json.load(open(osp.join(root, "meta_expressions.json")))
    jpeg_root = osp.join(root, "JPEGImages")
    ann_root = osp.join(root, "Annotations")
    has_ann = osp.isdir(ann_root)
    out: Dict[str, VideoData] = {}
    for vid, vinfo in meta["videos"].items():
        objs = vinfo["objects"]
        # frame list: union over objects (they usually share the same list)
        frames = sorted({f for o in objs.values() for f in o["frames"]})
        out[vid] = VideoData(
            video_id=vid,
            frame_names=frames,
            objects=objs,
            jpeg_dir=osp.join(jpeg_root, vid),
            ann_dir=osp.join(ann_root, vid) if has_ann else None,
        )
    return out


def load_rgb(path: str) -> np.ndarray:
    """Load an image as HxWx3 RGB uint8."""
    # Respect JPEG EXIF orientation. OpenCV (used by Mask R-CNN) may apply this
    # automatically, while plain PIL does not; ignoring it can rotate only the
    # refinement input and produce masks with transposed aspect ratios.
    with Image.open(path) as im:
        return np.array(ImageOps.exif_transpose(im).convert("RGB"))


def load_gt_mask(path: str, obj_id: int) -> np.ndarray:
    """Load palette PNG and return binary HxW uint8 for the given object id."""
    arr = np.array(Image.open(path))
    return (arr == obj_id).astype(np.uint8)


def mask_to_box(mask: np.ndarray) -> Optional[List[float]]:
    """Tight [x, y, w, h] bounding box of a binary mask, or None if empty."""
    ys, xs = np.where(mask > 0)
    if len(xs) == 0:
        return None
    x1, y1, x2, y2 = xs.min(), ys.min(), xs.max(), ys.max()
    return [float(x1), float(y1), float(x2 - x1 + 1), float(y2 - y1 + 1)]


# --------------------------------------------------------------------------- #
# (A) REAL Module-1 adapter: read proposals from MAttNet's JSON dump
# --------------------------------------------------------------------------- #
class MAttNetJsonProposals:
    """Adapter for the output of a (lightly extended) test_single_frame.py.

    Expected JSON schema (write this from your Module 1 over all frames):

      {
        "<video_id>": {
          "<obj_id>": {
            "<frame_name>": [
              {"box": [x,y,w,h], "score": <match_score>,
               "objectness": <maskrcnn_score>, "category_id": <int>,
               "det_id": <int>},
              ...
            ],
            ...
          }
        }
      }

    `score` is MAttNet's matching score (the bear's 0.286); `objectness` is the
    Mask R-CNN detection score (the per-det `score` in test_single_frame.py).
    """

    def __init__(self, json_path: str):
        self.data = json.load(open(json_path))

    def proposals(
        self, video: VideoData, obj_id: str
    ) -> List[Proposal]:
        vd = self.data[video.video_id][str(obj_id)]
        frame_to_ord = {n: i for i, n in enumerate(video.frame_names)}
        props: List[Proposal] = []
        for fname, dets in vd.items():
            f = frame_to_ord.get(fname)
            if f is None:
                continue
            for d in dets:
                props.append(Proposal(
                    box=d["box"], score=float(d["score"]),
                    objectness=float(d.get("objectness", 1.0)),
                    frame_idx=f, det_id=int(d.get("det_id", -1)),
                ))
        return props


# --------------------------------------------------------------------------- #
# (B) DEMO Module-1 simulator: GT mask -> jittered box + distractors
# --------------------------------------------------------------------------- #
class GTBoxProposals:
    """Simulate Module 1 from ground-truth masks (DEMO ONLY).

    For each frame we emit:
      * the true object's GT box, jittered +-`jitter`% (paper-style box noise),
        with a high matching score;
      * a few distractor boxes taken from OTHER objects' GT in the same frame,
        with lower (and occasionally spiking) matching scores, to make the
        temporal-consistency step do real work.

    This lets Module 2 + Module 3 run end-to-end without the MAttNet/MaskRCNN
    stack. The numbers are not a benchmark — they validate the plumbing.
    """

    def __init__(self, jitter: float = 0.15, n_distractors: int = 3,
                 seed: int = 0):
        self.jitter = jitter
        self.n_distractors = n_distractors
        self.rng = np.random.default_rng(seed)

    def _jitter_box(self, box: List[float]) -> List[float]:
        x, y, w, h = box
        jx = self.rng.uniform(-self.jitter, self.jitter) * w
        jy = self.rng.uniform(-self.jitter, self.jitter) * h
        jw = self.rng.uniform(-self.jitter, self.jitter) * w
        jh = self.rng.uniform(-self.jitter, self.jitter) * h
        return [x + jx, y + jy, max(4.0, w + jw), max(4.0, h + jh)]

    def proposals(self, video: VideoData, obj_id: str) -> List[Proposal]:
        assert video.ann_dir is not None, "GT proposals need Annotations/"
        oid = int(obj_id)
        other_ids = [int(k) for k in video.objects.keys() if int(k) != oid]
        props: List[Proposal] = []
        det = 0
        for f, fname in enumerate(video.frame_names):
            apath = video.ann_path(fname)
            if apath is None or not osp.isfile(apath):
                continue
            arr = np.array(Image.open(apath))

            # --- true object ---
            tmask = (arr == oid).astype(np.uint8)
            tbox = mask_to_box(tmask)
            if tbox is not None:
                props.append(Proposal(
                    box=self._jitter_box(tbox),
                    score=float(self.rng.uniform(0.55, 0.75)),
                    objectness=float(self.rng.uniform(0.85, 0.99)),
                    frame_idx=f, det_id=det)); det += 1

            # --- distractors from other objects ---
            self.rng.shuffle(other_ids)
            for od in other_ids[: self.n_distractors]:
                dmask = (arr == od).astype(np.uint8)
                dbox = mask_to_box(dmask)
                if dbox is None:
                    continue
                # distractors usually lower, but occasionally spike above the
                # true object in a single frame (that's what TC must fix)
                spike = self.rng.random() < 0.15
                sc = self.rng.uniform(0.6, 0.8) if spike else self.rng.uniform(0.3, 0.55)
                props.append(Proposal(
                    box=self._jitter_box(dbox), score=float(sc),
                    objectness=float(self.rng.uniform(0.7, 0.95)),
                    frame_idx=f, det_id=det)); det += 1
        return props
```

```python
# eval ytos script:
#!/usr/bin/env bash
# Evaluate the fine-tuned MAttNet + 5-channel refinement on Refer-YTVOS valid.
set -euo pipefail

SCRIPT_DIR=$(cd "$(dirname "${BASH_SOURCE[0]}")" && pwd)
REPO_ROOT=$(cd "${SCRIPT_DIR}/../.." && pwd)
cd "${REPO_ROOT}"

GPU_ID=${1:-0}
YTVOS_DIR=${2:-data/youtube-vos-2019}
VALID_DIR="${YTVOS_DIR}/valid"

DATASET_SPLITBY="ytvos_train"
MODEL_ID=${MODEL_ID:-mrcn_cmr_with_st}
CHECKPOINT="output/${DATASET_SPLITBY}/${MODEL_ID}.pth"
REFINEMENT_CKPT=${REFINEMENT_CKPT:-output/refinement_5ch/ytvos/ckpt_final.pth}

EVAL_STRIDE=${EVAL_STRIDE:-2}
EVAL_MAX_VIDEOS=${EVAL_MAX_VIDEOS:-9999}
EVAL_MAX_EXPRS=${EVAL_MAX_EXPRS:-9999}
EVAL_TOPK=${EVAL_TOPK:-1}
EVAL_TEMPORAL=${EVAL_TEMPORAL:-1}
EVAL_FLOW=${EVAL_FLOW:-1}

if [ ! -f "${CHECKPOINT}" ]; then
  echo "Error: MAttNet checkpoint not found: ${CHECKPOINT}" >&2
  exit 1
fi
if [ ! -f "${REFINEMENT_CKPT}" ]; then
  echo "Error: refinement checkpoint not found: ${REFINEMENT_CKPT}" >&2
  exit 1
fi
if [ ! -f "${VALID_DIR}/meta_expressions.json" ] || \
   [ ! -d "${VALID_DIR}/JPEGImages" ] || \
   [ ! -d "${VALID_DIR}/Annotations" ]; then
  echo "Error: incomplete Refer-YouTube-VOS valid split: ${VALID_DIR}" >&2
  exit 1
fi

VALID_ABS=$(cd "${VALID_DIR}" && pwd)
REFINEMENT_CKPT_ABS=$(cd "$(dirname "${REFINEMENT_CKPT}")" && pwd)/$(basename "${REFINEMENT_CKPT}")

echo "=============================================================="
echo " Refer-YouTube-VOS validation"
echo " MAttNet:    ${CHECKPOINT}"
echo " Refinement: ${REFINEMENT_CKPT_ABS}"
echo " Data:       ${VALID_ABS}"
echo "=============================================================="

(
  cd refvos
  CUDA_VISIBLE_DEVICES="${GPU_ID}" python eval_ytvos_jf.py \
    --valid_dir "${VALID_ABS}" \
    --dataset_splitBy "${DATASET_SPLITBY}" \
    --model_id "${MODEL_ID}" \
    --gt_layout perobj \
    --max_videos "${EVAL_MAX_VIDEOS}" \
    --max_exprs "${EVAL_MAX_EXPRS}" \
    --frame_stride "${EVAL_STRIDE}" \
    --topk "${EVAL_TOPK}" \
    --refinement_checkpoint "${REFINEMENT_CKPT_ABS}" \
    --temporal "${EVAL_TEMPORAL}" \
    --flow "${EVAL_FLOW}"
)
```
